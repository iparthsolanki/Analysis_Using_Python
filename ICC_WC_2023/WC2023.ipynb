```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a= pd.read_csv('c:/Users/solan/Downloads/CWC2023.csv')
```

```python
a.columns
```

```python
a.head()
```

```python
df = a[['Match ID','Match Date', 'City', 'Stadium', 'Team A','Boundaries A','Team B', 'Toss Winner', 'Toss Decision', 'Score A','Score B','Boundaries B','Wining Team','Man of the Match']]
```

```python
df.head()
```

```python
df.shape
```

```python
mc =df['City'].value_counts()
mc
```

```python
plt.figure(figsize=(12,6))
plt.bar(mc.index,mc.values,color=sns.color_palette('viridis'))
plt.xticks(rotation=45)
for i, v in enumerate (mc.values):
    plt.text(i, v + 0.1, str(v), ha='center', fontweight='bold')
plt.show()
```

```python
s = df['Stadium'].value_counts()
s
```

```python
plt.figure(figsize=(12,6))
plt.bar(s.index,s.values,color=sns.color_palette('mako'))
plt.xticks(rotation=90)
for i, v in enumerate (s.values):
    plt.text(i, v + 0.1, str(v), ha='center', fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(8,8))
plt.pie(df['Toss Decision'].value_counts(),labels=df['Toss Decision'].value_counts().index,autopct='%1.1f%%',colors=['red','blue'])
plt.show()
```

```python
teams = pd.concat([df['Team A'],df['Team B']])
```

```python
played = teams.value_counts()
played
```

```python
plt.figure(figsize=(12,6))
plt.bar(played.index,played.values,color=sns.color_palette('mako'))
for i, v in enumerate (played.values):
    plt.text(i, v + 0.1, str(v), ha='center', fontweight='bold')
plt.xticks(rotation=45)

plt.show()
```

```python
win = df['Wining Team'].value_counts()
win
```

```python

plt.figure(figsize=(12,8))


plt.bar(win.index, win.values, color=sns.color_palette('bright'))


plt.plot(win.index, win.values, 
         color='black', marker='o', linewidth=2, label='Trend')

plt.xticks(rotation=90)
for i, v in enumerate(win.values):
    plt.text(i, v + 0.1, str(v), ha='center', fontweight='bold')

plt.xlabel("Teams")
plt.ylabel("Wins")
plt.title("Team-wise Wins with Trend Line (WC 2023)")
plt.legend()
plt.show()
```

```python
mvp = df['Man of the Match'].value_counts()
mvp
```

```python
plt.figure(figsize=(18,8))


plt.bar(mvp.index[0:10], mvp.values[0:10],
        color=sns.color_palette("bright"))

for i, v in enumerate(mvp.values[0:10]):
    plt.text(i, v + 0.1, str(v), ha='center', fontweight='bold')
plt.xticks(rotation=90)
plt.title("Top 10 MVP Awards WC 2023 (with Trend)")
plt.ylabel("Count")
plt.show()
```

```python
score = pd.concat([df['Score A'], df['Score B']])
score
```

```python
plt.figure(figsize=(8,6))
sns.boxenplot(y=score, color='lightblue')
plt.title("Score Distribution WC 2023")
plt.ylabel("Scores")
plt.show()

```

```python
plt.figure(figsize=(10,6))
sns.histplot(score,bins=20,kde=True,color='green')
plt.title("Score Distribution Histogram WC 2023")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(x=score.index, y=score.values, hue=score.values, palette='coolwarm', s=100)
plt.title("Score Distribution Scatter Plot WC 2023")
plt.xlabel("Match Instances")
plt.ylabel("Scores")
plt.show()
```
